```python
import pandas as pd
import least_entropy_vectorized as epf
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv("ogle_collection/rrlyrae_phot/OGLE-LMC-RRLYR-00005.dat", sep=" ", names=["hjd", "I", "inc_I"])
```

```python
plt.scatter(data["hjd"], data["I"], s=2, c="k")
```

```python
p_candidates, _ = epf.find_best_period(data["hjd"],data["I"], 0.1,1.0,15000, n_candidates=6)
```

```python
for p in p_candidates:
    phases0 = epf.get_phases(data["hjd"], data["I"], p)
    phases0 = pd.concat([phases0, phases0 + 1])

    mag = data["I"]
    mag = pd.concat([mag, mag])

    print(f"Periodo {p}")
    plt.scatter(phases0, mag, s=3, c="k")
    plt.show()
```
