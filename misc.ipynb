```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("ogle_collection/classical_cefeids.txt",sep="\t", skiprows=6)
```

```python
df.head()
```

```python
x = 99.99  # percentil a calcular

percentil = np.percentile(df["P_1"], x)
print(f"El {x}% de las estrellas cefeidas tiene P_1 <= {percentil:.4f} días")
print(f"Período mínimo {df["P_1"].min()}, Período máximo {df["P_1"].max()} ")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.hist(df["P_1"])
ax1.set_title("Distribución de períodos en cefeidas clásicas")
ax1.set_xlabel("Período (días)")
ax1.set_ylabel("Cantidad")

ax2.hist(df["P_1"], log=True)
ax2.set_title("Distribución de períodos en cefeidas clásicas (log)")
ax2.set_xlabel("Período (días)")
ax2.set_ylabel("Cantidad (log)")

plt.tight_layout()
plt.show()
```

```python
plt.hist(df["I"], log=True, color="red")
plt.title("Distribución de magnitudes (I) en cefeidas clásicas")
plt.xlabel("Período")
```

```python
df = pd.read_csv("ogle_collection/binaries.txt",sep="\t", skiprows=6)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.hist(df["P"])
ax1.set_title("Distribución de períodos en estrellas binarias")
ax1.set_xlabel("Período (días)")
ax1.set_ylabel("Cantidad")

ax2.hist(df["P"], log=True)
ax2.set_title("Distribución de períodos en estrellas binarias (log)")
ax2.set_xlabel("Período (días)")
ax2.set_ylabel("Cantidad (log)")

plt.tight_layout()
plt.show()
```

```python
plt.hist(df["I"], log=True, color="red")
plt.title("Distribución de magnitudes (I) en binarias")
plt.xlabel("Período")
```

```python
x = 98  # percentil a calcular

percentil = np.percentile(df["P"], x)
print(f"El {x}% de las estrellas binarias tiene P_1 <= {percentil:.4f} días")
print(f"Período mínimo {df["P"].min()}, Período máximo {df["P"].max()} ")
```
